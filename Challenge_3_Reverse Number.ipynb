```python
import pandas as pd
import numpy as np

data= pd.read_excel("Challenge-3-Reverse Number.xlsx")
df=np.array(data).astype(str).flatten()

result= [x[::-1] for x in df]
result= pd.Series(result, name="Number")

print("-------Input Table-------")
print(df)
print()
print("-------Output Table------")
print (result)
```
